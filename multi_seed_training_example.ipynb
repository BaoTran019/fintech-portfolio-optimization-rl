## Performance Analysis

Analyze the metrics from different seeds to understand performance variability.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load metrics from different seeds
metrics_files = [
    'results/metrics/PPO_seed_42_metrics.csv',
    'results/metrics/PPO_seed_43_metrics.csv',
    'results/metrics/PPO_seed_44_metrics.csv'
]

metrics_list = []
for file in metrics_files:
    try:
        df = pd.read_csv(file)
        metrics_list.append(df)
    except FileNotFoundError:
        print(f"File {file} not found")

if metrics_list:
    # Combine metrics
    combined_metrics = pd.concat(metrics_list, ignore_index=True)
    print("Performance Statistics Across Seeds:")
    print(combined_metrics.describe())
    
    # Plot Sharpe ratios
    plt.figure(figsize=(8, 5))
    plt.bar(range(len(combined_metrics)), combined_metrics['Sharpe'])
    plt.xlabel('Seed Index')
    plt.ylabel('Sharpe Ratio')
    plt.title('Sharpe Ratio Across Different Seeds')
    plt.show()
else:
    print("No metrics files found. Run validation first.")
```
